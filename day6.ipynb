```python
import numpy as np
arr = np.array([1, np.nan, 5, np.inf])
print(arr)
print(np.isnan(arr))
print(np.isinf(arr))
```

```python
import numpy as np
arr = np.array([1, 5,])
print(np.isnan(arr))
print(np.isinf(arr))
```

```python
import numpy as np
arr = np.array([1, np.nan, 3, np.inf])
arr[np.isnan(arr)] = 0
arr[np.isinf(arr)] = 0
print(arr)
```

```python
import numpy as np
arr = np.array([1, np.nan, 3, np.inf])
arr[np.isnan(arr)] = 0
print(arr)
arr[np.isinf(arr)] = 0
print(arr)
```

```python
import numpy as np
arr = np.array([1, np.nan, 5, np.inf, 7])
clean_arr = arr[~np.isnan(arr)]
print(clean_arr)
clean_arr = clean_arr[~np.isinf(clean_arr)]
print(clean_arr)
```

```python
import numpy as np
arr = np.array([1, 2, 3])
print(arr)
copy_arr = arr.copy()
print(copy_arr)
copy_arr[0] = 2
print(copy_arr)
print(arr)
view_arr = arr.view()

view_arr[0] = 10
print(arr)
print(view_arr)
```
