```python
import numpy as np
arr = np.array([[25.5, 26.0, np.nan, 27.8, np.inf, 26.4]])
print(np.isnan(arr))
print(np.isinf(arr))
```

```python
arr[np.isnan(arr)] = 0
arr[np.isinf(arr)] = 0
print(arr)
```
