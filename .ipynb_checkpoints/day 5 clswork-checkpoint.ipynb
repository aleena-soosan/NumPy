```python
import numpy as np
arr = np.array([78, 85, 62, 90, 55, 88])
total = np.sum(arr)
print(f"total marks of the students:{total}")
```

```python
print(f"The average mark:{np.mean(arr)}")
```

```python
print(f"The median mark:{np.median(arr)}")
```

```python
print(f" The standard deviation:{np.std(arr)}")
```

```python
print(f"The minimum mark:{np.min(arr)}")
print(f"The maximum mark:{np.max(arr)}")
```

```python
scalar = 5
print(f"Total mark with grace mark:{arr + scalar}")
```

```python
print(f"Multiply all marks by 1.1 :{arr * 1.1}")
```
