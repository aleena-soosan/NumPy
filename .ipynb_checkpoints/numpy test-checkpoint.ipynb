```python
import numpy as np
import pandas as pd

customer_df = pd.read_excel("dataset.xlsx",sheet_name="customer_data.csv")
sales_df = pd.read_excel("dataset.xlsx",sheet_name="sales_data.csv")
noise_df = pd.read_excel("dataset.xlsx",sheet_name="noise_data.csv")
```

```python
customer = customer_df.to_numpy()

sales = sales_df.to_numpy()

noise = noise_df.to_numpy()
```

```python
print("CUSTOMER DATA")

print(customer.shape)

print("Rows =", customer.shape[0])

print("Columns =", customer.shape[1])

print("Total Values =", customer.size)

print("Dimensions =", customer.ndim)

print("Data Type =", customer.dtype)
```

```python
print("SALES DATASET")

print("Shape:", sales.shape)

print("Rows:", sales.shape[0])

print("Columns:", sales.shape[1])

print("Total Values:", sales.size)

print("Dimensions:", sales.ndim)

print("Data Type:", sales.dtype)

print()
```

```python
print("NOISE DATASET")

print("Shape:", noise.shape)

print("Rows:", noise.shape[0])

print("Columns:", noise.shape[1])

print("Total Values:", noise.size)

print("Dimensions:", noise.ndim)

print("Data Type:", noise.dtype)

print()
```
