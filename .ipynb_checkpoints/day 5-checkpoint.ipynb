```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.dot(a, b))

m1 = np.array([[1, 2],
               [3, 4]])
m2 = np.array([[5, 6],
               [7, 8]])
print(np.matmul(m1, m2))
```

```python
import numpy as np

a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(a @ b)
print(np.dot(a, b))
```

```python
import numpy as np

A = np.array([[2, 1], [1, 3]])
b = np.array([8, 13])
x = np.linalg.solve(A, b)
print(x)
```

```python
import numpy as np

a = np.array([[2, 0], [0, 3]])
values, vectors = np.linalg.eig(a)
print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
```
