```python
import numpy as np
arr = np.array( [250.75, 300.40, 275.60, 400.90, 150.30, 500.50, 350.80])
print(np.sum(arr))
print(np.mean(arr))
print(np.median(arr))
print(np.min(arr))
print(np.max(arr))

      
```

```python
print(np.floor(arr))
print(np.ceil(arr))
print(np.round(arr))

```

```python
import numpy as np
arr = np.array([1, 2, 3])
print(np.exp(arr))
```

```python
print(np.cumsum(arr))
```

```python
A = np.array([[2, 1],[3, 4]])
B = np.array([[1, 2],[5, 6]])
print(A @ B)
```

```python
print(np.linalg.det(A))
```

```python
print(np.linalg.inv(A))
```

```python
values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)
print("Eigenvectors:\n", vectors)
```

```python
A = np.array([[2,1],[3,4]])
B = np.array([5,11])
x = np.linalg.solve(A, B)
print(x)
```
