```python
import numpy as np
data = np.array([10, 20, 30, 40, 50])
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))
print(normalized)
```

```python
import numpy as np
data = np.array([1, 2, 3, 4, 5])
print("Mean:", np.mean(data))
print("Std Dev:", np.std(data))
```

```python
import numpy as np
import sys
list_obj = [1, 2, 3, 4]
array_obj = np.array([1, 2, 3, 4])
print(sys.getsizeof(list_obj))       # Size of list
print(array_obj.nbytes)              # Size of NumPy array
```
