```python
import numpy as np
marks = np.array([35, 55, 75, 40, 90])
print(f"array:\n{marks}")
```

```python
result = np.where(marks >= 40, 'pass', 'fail')
print(result)
```

```python
conditions = [marks < 40, marks < 60,marks < 80]
choices = ['fail', 'C','B']
result = np.select(conditions, choices, default='A')
print(result)
```

```python
marks_dup = np.array([50, 60, 50, 70, 60, 80])
print(np.unique(marks_dup))
```

```python
marks = np.array([35, 55, 75, 40, 90])
print(np.any(marks < 40))
```

```python
print(np.all(marks >= 40))
```
