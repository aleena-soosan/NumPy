```python
import numpy as np
arr = np.array([25, 45, 65, 85, 95])
score = np.where(arr < 40,'poor','good')
print(score)
```

```python
conditions = [arr < 40,arr < 70,arr < 90]
choice = ['poor','avg','good']
result = np.select(conditions,choice,default='execellent')
print(result)
```
