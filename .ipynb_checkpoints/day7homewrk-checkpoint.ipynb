```python
import numpy as np
arr = np.array([45, 55, 60, 70, 85])
arr_1 = 5
result = arr + arr_1
print(result)
```

```python
normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
print(normalized)
```

```python
np.savetxt('arr.txt',arr, delimiter=',')
```

```python
arr = np.genfromtxt('arr.txt', delimiter=',')
print(arr)
```
