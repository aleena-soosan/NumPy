```python
#array printing
import numpy as np
arr1D = np.array([60,70,80,90,100])
arr2D = np.array([[80,40,50],[50,40,70]])
arr3D = np.array([[[40,50,30],[60,40,50]],[[40,70,60],[50,30,60]]])
print(f"Marks of 5 students: {arr1D}")
print(f"Marks of 2 students in 3 sub:{arr2D}")
print(f"Marks of 2 classess with 2 students and 3 sub:{arr3D}")    
```

```python
print(f"shape of array:{arr2D.shape}")
print(f"number of dimention of array:{arr2D.ndim}")
print(f"data type of array:{arr2D.dtype}")
print(f"size of array:{arr2D.size}")
```

```python
import numpy as np
arr = np.zeros((3,4))
arr1 = np.ones((2,3))
arr2 = np.full((3,3),5)
print(f"array of zeros:{arr}")
print(f"array of ones:{arr1}")
print(f"array filled with 5:{arr2}")
```

```python
import numpy as np
arr = np.arange(10,50,10)
arr1 = np.linspace(0,100,5)
print(f"Array from 10 to 50 with step 10:{arr}")
print(f"Array:{arr1}")
```
