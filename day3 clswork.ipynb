```python
import numpy as np
marks_term1 = np.array([60, 70, 80, 90])
marks_term2 = np.array([65, 75, 85, 95])
total_mark = marks_term1 + marks_term2
print(f"Total marks:\n{total_mark}")
```

```python
difference = marks_term2 - marks_term1
print(f"The difference :{difference}")
```

```python
increased_mark = marks_term1 * 1.05
print(f"The increased mark:{increased_mark}")
```

```python
increased_mark = marks_term2 * 1.05
print(f"The increased mark:{increased_mark}")
```

```python
print(f"marks greater than 150:\n{total_mark > 150}")
```

```python
print(f"marks greater than 150:\n{total_mark[total_mark > 150]}")
```

```python
marks = total_mark[((total_mark >= 140) & (total_mark <= 170))]
print(marks)
```

```python
result = total_mark.sum()
print(result)
```

```python
result = total_mark.mean()
print(result)
```

```python
result = total_mark.max()
print(result)
```

```python
result = total_mark.min()
print(result)
```

```python
result = total_mark.std()
print(result)
```
