```python
import numpy as np
attendance = (np.random.rand(10))
print(attendance)
```

```python
present = attendance[attendance > 0.5]
print(present)
```

```python
absent = attendance[attendance <= 0.5]
print(absent)
```

```python
scores =  np.random.randint(0,101,15)
print(scores)
```

```python
print("\nScores divisible by 5:")
print(scores[scores > 60])
```

```python
print("\nScores divisible by 5:")
print(scores[scores % 5 == 0])
```
